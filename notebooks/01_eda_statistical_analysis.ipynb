```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv("../data/processed/customer_features.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Churn Counts:")
print(df["churn"].value_counts())

print("\nChurn Percentage:")
churn_percentage = df["churn"].value_counts(normalize=True) * 100
print(churn_percentage)
```

Correlation Analysis

```python
# Calculate correlation with churn
correlation = df.corr(numeric_only=True)

churn_correlation = (
    correlation["churn"]
    .sort_values(ascending=False)
)

print("Correlation of Features with Churn:")
print(churn_correlation)
```

Hypothesis Testing

```python
# Compare tenure between churned and non-churned customers

churned_tenure = df[df["churn"] == 1]["tenure_months"]
non_churned_tenure = df[df["churn"] == 0]["tenure_months"]

t_stat, p_value = stats.ttest_ind(
    churned_tenure,
    non_churned_tenure,
    equal_var=False
)

print("Churned Customer Average Tenure:",
      round(churned_tenure.mean(), 2))

print("Non-Churned Customer Average Tenure:",
      round(non_churned_tenure.mean(), 2))

print("T-statistic:", round(t_stat, 4))
print("P-value:", p_value)

if p_value < 0.05:
    print("\nResult: Statistically significant difference detected.")
else:
    print("\nResult: No statistically significant difference detected.")
```

Analyze Support Tickets

```python
# Compare support tickets between churned and non-churned customers

churned_tickets = df[df["churn"] == 1]["support_tickets"]
non_churned_tickets = df[df["churn"] == 0]["support_tickets"]

t_stat, p_value = stats.ttest_ind(
    churned_tickets,
    non_churned_tickets,
    equal_var=False
)

print(
    "Churned Customer Average Support Tickets:",
    round(churned_tickets.mean(), 2)
)

print(
    "Non-Churned Customer Average Support Tickets:",
    round(non_churned_tickets.mean(), 2)
)

print("T-statistic:", round(t_stat, 4))
print("P-value:", p_value)

if p_value < 0.05:
    print("\nResult: Statistically significant difference detected.")
else:
    print("\nResult: No statistically significant difference detected.")
```
